```python
import os
import json
import pickle
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report,
    precision_score,
    recall_score,
    f1_score
)

from tensorflow.keras.models import Sequential, load_model
from tensorflow.keras.layers import Dense, Dropout, Input
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.utils import to_categorical


warnings.filterwarnings("ignore")


# =====================================================================
# 2. PROJECT PATH
# =====================================================================

BASE_DIR = r"C:\Users\sagni\Downloads\AI-Based Vegetable Production Pattern Analysis"

CSV_PATH = os.path.join(
    BASE_DIR,
    "Production_of_Important_Vegetable_crops_In_Tonnes_2019_20.csv"
)

os.makedirs(BASE_DIR, exist_ok=True)


# =====================================================================
# 3. OUTPUT FILE PATHS
# =====================================================================

ACCURACY_GRAPH_PATH = os.path.join(
    BASE_DIR,
    "accuracy_graph.png"
)

HEATMAP_PATH = os.path.join(
    BASE_DIR,
    "confusion_matrix_heatmap.png"
)

COMPARISON_GRAPH_PATH = os.path.join(
    BASE_DIR,
    "model_comparison_graph.png"
)

COMPARISON_CSV_PATH = os.path.join(
    BASE_DIR,
    "model_comparison.csv"
)

FEATURE_IMPORTANCE_GRAPH_PATH = os.path.join(
    BASE_DIR,
    "feature_importance_graph.png"
)

RESULT_CSV_PATH = os.path.join(
    BASE_DIR,
    "result.csv"
)

RESULT_GRAPH_PATH = os.path.join(
    BASE_DIR,
    "result_graph.png"
)

PREDICTION_CSV_PATH = os.path.join(
    BASE_DIR,
    "prediction.csv"
)

PREDICTION_GRAPH_PATH = os.path.join(
    BASE_DIR,
    "prediction_graph.png"
)


print("=" * 80)
print("AI-BASED VEGETABLE PRODUCTION PATTERN ANALYSIS")
print("=" * 80)


# =====================================================================
# 4. LOAD DATASET
# =====================================================================

df = pd.read_csv(CSV_PATH)

df.columns = df.columns.str.strip()

print("\nDataset successfully loaded.")
print("Dataset shape:", df.shape)

print("\nDataset columns:")

for col in df.columns:
    print("-", col)


# =====================================================================
# 5. IDENTIFY TALUK AND CROP COLUMNS
# =====================================================================

TALUK_COLUMN = "Taluks"


if TALUK_COLUMN not in df.columns:

    raise ValueError(
        f"Column '{TALUK_COLUMN}' was not found in the dataset."
    )


crop_columns = [
    col
    for col in df.columns
    if col != TALUK_COLUMN
]


print("\nCrop columns:")

for col in crop_columns:
    print("-", col)


# =====================================================================
# 6. CONVERT CROP COLUMNS TO NUMERIC
# =====================================================================

for col in crop_columns:

    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )


# =====================================================================
# 7. HANDLE MISSING VALUES
# =====================================================================

for col in crop_columns:

    median_value = df[col].median()

    df[col] = df[col].fillna(
        median_value
    )


# Negative production does not make sense.
for col in crop_columns:

    df[col] = df[col].clip(
        lower=0
    )


print("\nMissing values after cleaning:")

print(
    df[crop_columns]
    .isnull()
    .sum()
)


# =====================================================================
# 8. CALCULATE TOTAL VEGETABLE PRODUCTION
# =====================================================================

df["Total_Vegetable_Production"] = df[
    crop_columns
].sum(axis=1)


print("\nTotal vegetable production calculated.")


# =====================================================================
# 9. CREATE LOW / MEDIUM / HIGH TARGET
# =====================================================================

LOW_THRESHOLD = df[
    "Total_Vegetable_Production"
].quantile(0.33)


MEDIUM_THRESHOLD = df[
    "Total_Vegetable_Production"
].quantile(0.66)


print("\nProduction thresholds:")

print(
    "Low maximum:",
    round(LOW_THRESHOLD, 2)
)

print(
    "Medium maximum:",
    round(MEDIUM_THRESHOLD, 2)
)


def production_category(value):

    if value <= LOW_THRESHOLD:

        return "Low"

    elif value <= MEDIUM_THRESHOLD:

        return "Medium"

    else:

        return "High"


df["Production_Category"] = df[
    "Total_Vegetable_Production"
].apply(
    production_category
)


print("\nProduction category distribution:")

print(
    df["Production_Category"]
    .value_counts()
)


# =====================================================================
# 10. INPUT AND OUTPUT FEATURES
# =====================================================================

# Important:
# Total_Vegetable_Production is NOT included as X because the target itself
# was constructed from total vegetable production.
# Including it would create target leakage.

X = df[
    crop_columns
].copy()


y = df[
    "Production_Category"
].copy()


# =====================================================================
# 11. LABEL ENCODING
# =====================================================================

label_encoder = LabelEncoder()

y_encoded = label_encoder.fit_transform(
    y
)


print("\nClass mapping:")

for index, class_name in enumerate(
    label_encoder.classes_
):

    print(
        index,
        "=",
        class_name
    )


# =====================================================================
# 12. TRAIN / TEST SPLIT
# =====================================================================

(
    X_train,
    X_test,
    y_train,
    y_test
) = train_test_split(

    X,

    y_encoded,

    test_size=0.20,

    random_state=42,

    stratify=y_encoded
)


print("\nTraining samples:", len(X_train))

print(
    "Testing samples:",
    len(X_test)
)


# =====================================================================
# 13. RANDOM FOREST
# =====================================================================

print("\nTraining Random Forest...")


rf_model = RandomForestClassifier(

    n_estimators=300,

    random_state=42,

    class_weight="balanced"
)


rf_model.fit(
    X_train,
    y_train
)


rf_train_prediction = rf_model.predict(
    X_train
)


rf_test_prediction = rf_model.predict(
    X_test
)


RF_TRAIN_ACCURACY = accuracy_score(
    y_train,
    rf_train_prediction
)


RF_TEST_ACCURACY = accuracy_score(
    y_test,
    rf_test_prediction
)


RF_PRECISION = precision_score(

    y_test,

    rf_test_prediction,

    average="weighted",

    zero_division=0
)


RF_RECALL = recall_score(

    y_test,

    rf_test_prediction,

    average="weighted",

    zero_division=0
)


RF_F1 = f1_score(

    y_test,

    rf_test_prediction,

    average="weighted",

    zero_division=0
)


print(
    "\nRandom Forest Training Accuracy:",
    round(
        RF_TRAIN_ACCURACY * 100,
        2
    ),
    "%"
)


print(
    "Random Forest Testing Accuracy:",
    round(
        RF_TEST_ACCURACY * 100,
        2
    ),
    "%"
)


print("\nRandom Forest Classification Report:")


print(
    classification_report(

        y_test,

        rf_test_prediction,

        target_names=label_encoder.classes_,

        zero_division=0
    )
)


# =====================================================================
# 14. STANDARDIZE DATA FOR NEURAL NETWORK
# =====================================================================

scaler = StandardScaler()


X_train_scaled = scaler.fit_transform(
    X_train
)


X_test_scaled = scaler.transform(
    X_test
)


number_of_classes = len(
    label_encoder.classes_
)


y_train_nn = to_categorical(

    y_train,

    num_classes=number_of_classes
)


y_test_nn = to_categorical(

    y_test,

    num_classes=number_of_classes
)


# =====================================================================
# 15. BUILD NEURAL NETWORK
# =====================================================================

nn_model = Sequential([

    Input(
        shape=(
            X_train_scaled.shape[1],
        )
    ),

    Dense(
        64,
        activation="relu"
    ),

    Dropout(
        0.20
    ),

    Dense(
        32,
        activation="relu"
    ),

    Dropout(
        0.10
    ),

    Dense(
        16,
        activation="relu"
    ),

    Dense(
        number_of_classes,
        activation="softmax"
    )
])


nn_model.compile(

    optimizer="adam",

    loss="categorical_crossentropy",

    metrics=[
        "accuracy"
    ]
)


early_stopping = EarlyStopping(

    monitor="val_loss",

    patience=15,

    restore_best_weights=True
)


print("\nTraining Neural Network...")


history = nn_model.fit(

    X_train_scaled,

    y_train_nn,

    validation_split=0.20,

    epochs=150,

    batch_size=16,

    callbacks=[
        early_stopping
    ],

    verbose=1
)


# =====================================================================
# 16. NEURAL NETWORK PREDICTIONS
# =====================================================================

nn_train_probability = nn_model.predict(
    X_train_scaled,
    verbose=0
)


nn_test_probability = nn_model.predict(
    X_test_scaled,
    verbose=0
)


nn_train_prediction = np.argmax(
    nn_train_probability,
    axis=1
)


nn_test_prediction = np.argmax(
    nn_test_probability,
    axis=1
)


NN_TRAIN_ACCURACY = accuracy_score(
    y_train,
    nn_train_prediction
)


NN_TEST_ACCURACY = accuracy_score(
    y_test,
    nn_test_prediction
)


NN_PRECISION = precision_score(

    y_test,

    nn_test_prediction,

    average="weighted",

    zero_division=0
)


NN_RECALL = recall_score(

    y_test,

    nn_test_prediction,

    average="weighted",

    zero_division=0
)


NN_F1 = f1_score(

    y_test,

    nn_test_prediction,

    average="weighted",

    zero_division=0
)


print(
    "\nNeural Network Training Accuracy:",
    round(
        NN_TRAIN_ACCURACY * 100,
        2
    ),
    "%"
)


print(
    "Neural Network Testing Accuracy:",
    round(
        NN_TEST_ACCURACY * 100,
        2
    ),
    "%"
)


# =====================================================================
# 17. ACCURACY GRAPH
# =====================================================================
# Shows training and validation accuracy across epochs.
# =====================================================================

plt.figure(
    figsize=(
        10,
        6
    )
)


plt.plot(
    history.history["accuracy"],
    label="Training Accuracy",
    linewidth=2
)


plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy",
    linewidth=2
)


plt.xlabel(
    "Epoch"
)


plt.ylabel(
    "Accuracy"
)


plt.title(
    "Neural Network Training and Validation Accuracy"
)


plt.legend()


plt.grid(
    alpha=0.3
)


plt.tight_layout()


plt.savefig(
    ACCURACY_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


print(
    "\nSaved:",
    ACCURACY_GRAPH_PATH
)


# =====================================================================
# 18. CONFUSION MATRIX HEATMAP
# =====================================================================

cm = confusion_matrix(

    y_test,

    rf_test_prediction
)


plt.figure(
    figsize=(
        8,
        6
    )
)


plt.imshow(
    cm,
    interpolation="nearest"
)


plt.title(
    "Random Forest Confusion Matrix"
)


plt.colorbar()


tick_marks = np.arange(
    len(
        label_encoder.classes_
    )
)


plt.xticks(
    tick_marks,
    label_encoder.classes_
)


plt.yticks(
    tick_marks,
    label_encoder.classes_
)


threshold = cm.max() / 2


for i in range(
    cm.shape[0]
):

    for j in range(
        cm.shape[1]
    ):

        plt.text(

            j,

            i,

            str(
                cm[i, j]
            ),

            horizontalalignment="center",

            verticalalignment="center",

            color="white"
            if cm[i, j] > threshold
            else "black",

            fontsize=12
        )


plt.ylabel(
    "Actual Production Category"
)


plt.xlabel(
    "Predicted Production Category"
)


plt.tight_layout()


plt.savefig(
    HEATMAP_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


print(
    "Saved:",
    HEATMAP_PATH
)


# =====================================================================
# 19. MODEL COMPARISON DATA
# =====================================================================

comparison_df = pd.DataFrame({

    "Model": [
        "Random Forest",
        "Neural Network"
    ],

    "Accuracy": [
        RF_TEST_ACCURACY,
        NN_TEST_ACCURACY
    ],

    "Precision": [
        RF_PRECISION,
        NN_PRECISION
    ],

    "Recall": [
        RF_RECALL,
        NN_RECALL
    ],

    "F1_Score": [
        RF_F1,
        NN_F1
    ]
})


comparison_df.to_csv(
    COMPARISON_CSV_PATH,
    index=False
)


print("\nModel comparison:")

print(
    comparison_df
)


# =====================================================================
# 20. MODEL COMPARISON GRAPH
# =====================================================================

metrics = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1_Score"
]


x = np.arange(
    len(metrics)
)


width = 0.35


plt.figure(
    figsize=(
        11,
        6
    )
)


plt.bar(

    x - width / 2,

    [
        RF_TEST_ACCURACY,
        RF_PRECISION,
        RF_RECALL,
        RF_F1
    ],

    width,

    label="Random Forest"
)


plt.bar(

    x + width / 2,

    [
        NN_TEST_ACCURACY,
        NN_PRECISION,
        NN_RECALL,
        NN_F1
    ],

    width,

    label="Neural Network"
)


plt.xticks(
    x,
    metrics
)


plt.ylim(
    0,
    1.10
)


plt.ylabel(
    "Score"
)


plt.title(
    "Machine Learning Model Performance Comparison"
)


plt.legend()


plt.grid(
    axis="y",
    alpha=0.3
)


# Add score values over bars.

for index, value in enumerate([

    RF_TEST_ACCURACY,
    RF_PRECISION,
    RF_RECALL,
    RF_F1

]):

    plt.text(

        index - width / 2,

        value + 0.02,

        f"{value:.3f}",

        ha="center"
    )


for index, value in enumerate([

    NN_TEST_ACCURACY,
    NN_PRECISION,
    NN_RECALL,
    NN_F1

]):

    plt.text(

        index + width / 2,

        value + 0.02,

        f"{value:.3f}",

        ha="center"
    )


plt.tight_layout()


plt.savefig(
    COMPARISON_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


print(
    "Saved:",
    COMPARISON_GRAPH_PATH
)


# =====================================================================
# 21. FEATURE IMPORTANCE GRAPH
# =====================================================================

feature_importance_df = pd.DataFrame({

    "Feature":
        crop_columns,

    "Importance":
        rf_model.feature_importances_
})


feature_importance_df = (
    feature_importance_df
    .sort_values(
        "Importance",
        ascending=True
    )
)


plt.figure(
    figsize=(
        10,
        7
    )
)


plt.barh(

    feature_importance_df[
        "Feature"
    ],

    feature_importance_df[
        "Importance"
    ]
)


plt.xlabel(
    "Feature Importance"
)


plt.ylabel(
    "Vegetable Crop"
)


plt.title(
    "Vegetable Production Feature Importance"
)


plt.grid(
    axis="x",
    alpha=0.3
)


plt.tight_layout()


plt.savefig(
    FEATURE_IMPORTANCE_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


print(
    "Saved:",
    FEATURE_IMPORTANCE_GRAPH_PATH
)


# =====================================================================
# 22. CREATE RESULT.CSV
# =====================================================================
# Result CSV contains test samples only.
# =====================================================================

test_indices = X_test.index


actual_categories = label_encoder.inverse_transform(
    y_test
)


rf_predicted_categories = label_encoder.inverse_transform(
    rf_test_prediction
)


nn_predicted_categories = label_encoder.inverse_transform(
    nn_test_prediction
)


result_df = pd.DataFrame({

    "Taluk":
        df.loc[
            test_indices,
            TALUK_COLUMN
        ].values,

    "Total_Vegetable_Production":
        df.loc[
            test_indices,
            "Total_Vegetable_Production"
        ].values,

    "Actual_Category":
        actual_categories,

    "Random_Forest_Prediction":
        rf_predicted_categories,

    "Neural_Network_Prediction":
        nn_predicted_categories,

    "RF_Correct":
        (
            actual_categories
            ==
            rf_predicted_categories
        ),

    "NN_Correct":
        (
            actual_categories
            ==
            nn_predicted_categories
        )
})


result_df.to_csv(
    RESULT_CSV_PATH,
    index=False
)


print("\nresult.csv:")

print(
    result_df.head(
        20
    )
)


print(
    "\nSaved:",
    RESULT_CSV_PATH
)


# =====================================================================
# 23. RESULT GRAPH
# =====================================================================
# Compare total production with production category.
# =====================================================================

result_graph_df = (
    result_df
    .sort_values(
        "Total_Vegetable_Production",
        ascending=False
    )
)


plt.figure(
    figsize=(
        14,
        7
    )
)


plt.bar(

    result_graph_df[
        "Taluk"
    ],

    result_graph_df[
        "Total_Vegetable_Production"
    ]
)


plt.xlabel(
    "Taluk"
)


plt.ylabel(
    "Total Vegetable Production (Tonnes)"
)


plt.title(
    "Vegetable Production Results for Test Taluks"
)


plt.xticks(
    rotation=90
)


plt.grid(
    axis="y",
    alpha=0.3
)


plt.tight_layout()


plt.savefig(
    RESULT_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


print(
    "Saved:",
    RESULT_GRAPH_PATH
)


# =====================================================================
# 24. PREDICT ALL TALUKS
# =====================================================================

all_rf_predictions = rf_model.predict(
    X
)


all_rf_probabilities = rf_model.predict_proba(
    X
)


all_scaled = scaler.transform(
    X
)


all_nn_probabilities = nn_model.predict(
    all_scaled,
    verbose=0
)


all_nn_predictions = np.argmax(
    all_nn_probabilities,
    axis=1
)


all_rf_categories = label_encoder.inverse_transform(
    all_rf_predictions
)


all_nn_categories = label_encoder.inverse_transform(
    all_nn_predictions
)


actual_all_categories = df[
    "Production_Category"
].values


# =====================================================================
# 25. RANDOM FOREST CONFIDENCE
# =====================================================================

rf_confidence = np.max(

    all_rf_probabilities,

    axis=1

) * 100


# =====================================================================
# 26. NEURAL NETWORK CONFIDENCE
# =====================================================================

nn_confidence = np.max(

    all_nn_probabilities,

    axis=1

) * 100


# =====================================================================
# 27. CREATE PREDICTION.CSV
# =====================================================================

prediction_df = pd.DataFrame({

    "Taluk":
        df[
            TALUK_COLUMN
        ].values,

    "Total_Vegetable_Production":
        df[
            "Total_Vegetable_Production"
        ].values,

    "Actual_Category":
        actual_all_categories,

    "RF_Predicted_Category":
        all_rf_categories,

    "RF_Confidence_Percentage":
        rf_confidence,

    "NN_Predicted_Category":
        all_nn_categories,

    "NN_Confidence_Percentage":
        nn_confidence
})


# Add crop production columns to prediction file.

for col in crop_columns:

    prediction_df[
        col
    ] = df[
        col
    ].values


prediction_df.to_csv(
    PREDICTION_CSV_PATH,
    index=False
)


print("\nprediction.csv:")

print(
    prediction_df.head(
        20
    )
)


print(
    "\nSaved:",
    PREDICTION_CSV_PATH
)


# =====================================================================
# 28. PREDICTION GRAPH
# =====================================================================
# To prevent the graph from becoming unreadable with all 228 taluks,
# show prediction confidence for the top 30 production taluks.
# prediction.csv still contains ALL taluks.
# =====================================================================

top_prediction_df = (

    prediction_df

    .sort_values(
        "Total_Vegetable_Production",
        ascending=False
    )

    .head(
        30
    )
)


x = np.arange(
    len(
        top_prediction_df
    )
)


width = 0.40


plt.figure(
    figsize=(
        16,
        8
    )
)


plt.bar(

    x - width / 2,

    top_prediction_df[
        "RF_Confidence_Percentage"
    ],

    width,

    label="Random Forest Confidence"
)


plt.bar(

    x + width / 2,

    top_prediction_df[
        "NN_Confidence_Percentage"
    ],

    width,

    label="Neural Network Confidence"
)


plt.xticks(

    x,

    top_prediction_df[
        "Taluk"
    ],

    rotation=90
)


plt.xlabel(
    "Taluk"
)


plt.ylabel(
    "Prediction Confidence (%)"
)


plt.title(
    "Prediction Confidence for Top 30 Vegetable Producing Taluks"
)


plt.ylim(
    0,
    105
)


plt.legend()


plt.grid(
    axis="y",
    alpha=0.3
)


plt.tight_layout()


plt.savefig(
    PREDICTION_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


print(
    "Saved:",
    PREDICTION_GRAPH_PATH
)


# =====================================================================
# 29. CATEGORY DISTRIBUTION GRAPH
# =====================================================================

category_counts = (
    prediction_df[
        "RF_Predicted_Category"
    ]
    .value_counts()
)


plt.figure(
    figsize=(
        8,
        6
    )
)


plt.bar(

    category_counts.index,

    category_counts.values
)


plt.xlabel(
    "Production Category"
)


plt.ylabel(
    "Number of Taluks"
)


plt.title(
    "Predicted Vegetable Production Category Distribution"
)


plt.grid(
    axis="y",
    alpha=0.3
)


CATEGORY_GRAPH_PATH = os.path.join(

    BASE_DIR,

    "production_category_distribution.png"
)


plt.tight_layout()


plt.savefig(
    CATEGORY_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


print(
    "Saved:",
    CATEGORY_GRAPH_PATH
)


# =====================================================================
# 30. SAVE MODELS
# =====================================================================

RF_MODEL_PATH = os.path.join(
    BASE_DIR,
    "vegetable_rf_model.pkl"
)


SCALER_PATH = os.path.join(
    BASE_DIR,
    "scaler.pkl"
)


ENCODER_PATH = os.path.join(
    BASE_DIR,
    "label_encoder.pkl"
)


NN_MODEL_PATH = os.path.join(
    BASE_DIR,
    "vegetable_nn_model.h5"
)


with open(
    RF_MODEL_PATH,
    "wb"
) as file:

    pickle.dump(
        rf_model,
        file
    )


with open(
    SCALER_PATH,
    "wb"
) as file:

    pickle.dump(
        scaler,
        file
    )


with open(
    ENCODER_PATH,
    "wb"
) as file:

    pickle.dump(
        label_encoder,
        file
    )


nn_model.save(
    NN_MODEL_PATH
)


# =====================================================================
# 31. FINAL PROJECT SUMMARY
# =====================================================================

print("\n")
print("=" * 80)

print(
    "PROJECT COMPLETED SUCCESSFULLY"
)

print("=" * 80)


print(
    "\nRANDOM FOREST"
)


print(
    "Training Accuracy:",
    f"{RF_TRAIN_ACCURACY * 100:.2f}%"
)


print(
    "Testing Accuracy:",
    f"{RF_TEST_ACCURACY * 100:.2f}%"
)


print(
    "Precision:",
    f"{RF_PRECISION * 100:.2f}%"
)


print(
    "Recall:",
    f"{RF_RECALL * 100:.2f}%"
)


print(
    "F1 Score:",
    f"{RF_F1 * 100:.2f}%"
)


print(
    "\nNEURAL NETWORK"
)


print(
    "Training Accuracy:",
    f"{NN_TRAIN_ACCURACY * 100:.2f}%"
)


print(
    "Testing Accuracy:",
    f"{NN_TEST_ACCURACY * 100:.2f}%"
)


print(
    "Precision:",
    f"{NN_PRECISION * 100:.2f}%"
)


print(
    "Recall:",
    f"{NN_RECALL * 100:.2f}%"
)


print(
    "F1 Score:",
    f"{NN_F1 * 100:.2f}%"
)


print("\nGenerated files:")


generated_files = [

    ACCURACY_GRAPH_PATH,

    HEATMAP_PATH,

    COMPARISON_GRAPH_PATH,

    COMPARISON_CSV_PATH,

    FEATURE_IMPORTANCE_GRAPH_PATH,

    RESULT_CSV_PATH,

    RESULT_GRAPH_PATH,

    PREDICTION_CSV_PATH,

    PREDICTION_GRAPH_PATH,

    CATEGORY_GRAPH_PATH,

    RF_MODEL_PATH,

    NN_MODEL_PATH,

    SCALER_PATH,

    ENCODER_PATH
]


for file in generated_files:

    print(
        "✓",
        os.path.basename(
            file
        )
    )


print(
    "\nAll outputs saved in:"
)


print(
    BASE_DIR
)


print(
    "\nDONE!"
)
```
