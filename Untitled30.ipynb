```python
import os
import json
import pickle
import warnings

import numpy as np
import pandas as pd
import yaml

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Input
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.utils import to_categorical

warnings.filterwarnings("ignore")


# ------------------------------------------------
# 2. PATHS
# ------------------------------------------------

BASE_DIR = r"C:\Users\sagni\Downloads\AI-Based Vegetable Production Pattern Analysis"

CSV_PATH = os.path.join(
    BASE_DIR,
    "Production_of_Important_Vegetable_crops_In_Tonnes_2019_20.csv"
)

os.makedirs(BASE_DIR, exist_ok=True)

print("=" * 70)
print("AI-BASED VEGETABLE PRODUCTION PATTERN ANALYSIS")
print("=" * 70)

print("\nProject folder:")
print(BASE_DIR)

print("\nDataset:")
print(CSV_PATH)


# ------------------------------------------------
# 3. LOAD DATASET
# ------------------------------------------------

df = pd.read_csv(CSV_PATH)

print("\nDataset loaded successfully.")
print("Dataset Shape:", df.shape)

print("\nColumns:")
for col in df.columns:
    print("-", col)

print("\nFirst 5 rows:")
print(df.head())


# ------------------------------------------------
# 4. DATASET INFORMATION
# ------------------------------------------------

print("\nDataset Information:")
print(df.info())

print("\nMissing Values:")
print(df.isnull().sum())


# ------------------------------------------------
# 5. CLEAN COLUMN NAMES
# ------------------------------------------------

df.columns = df.columns.str.strip()

print("\nCleaned columns:")
print(df.columns.tolist())


# ------------------------------------------------
# 6. IDENTIFY TALUK COLUMN
# ------------------------------------------------

TALUK_COLUMN = "Taluks"

if TALUK_COLUMN not in df.columns:
    raise ValueError(
        f"'{TALUK_COLUMN}' column could not be found in the dataset."
    )


# ------------------------------------------------
# 7. CONVERT CROP COLUMNS TO NUMERIC
# ------------------------------------------------

crop_columns = [
    col for col in df.columns
    if col != TALUK_COLUMN
]

for col in crop_columns:

    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )


# ------------------------------------------------
# 8. HANDLE MISSING VALUES
# ------------------------------------------------

for col in crop_columns:

    median_value = df[col].median()

    df[col] = df[col].fillna(median_value)


print("\nMissing values after preprocessing:")
print(df.isnull().sum())


# ------------------------------------------------
# 9. REMOVE NEGATIVE PRODUCTION VALUES
# ------------------------------------------------

for col in crop_columns:

    df[col] = df[col].clip(lower=0)


# ------------------------------------------------
# 10. CALCULATE TOTAL VEGETABLE PRODUCTION
# ------------------------------------------------

df["Total_Vegetable_Production"] = df[crop_columns].sum(axis=1)

print("\nTotal Vegetable Production calculated.")

print(
    df[
        [
            TALUK_COLUMN,
            "Total_Vegetable_Production"
        ]
    ].head()
)


# ------------------------------------------------
# 11. CREATE PRODUCTION CATEGORY
# ------------------------------------------------
# Production levels are determined using terciles.
#
# Bottom 33%  -> Low
# Middle 33%  -> Medium
# Top 33%     -> High
# ------------------------------------------------

q1 = df["Total_Vegetable_Production"].quantile(0.33)
q2 = df["Total_Vegetable_Production"].quantile(0.66)

print("\nProduction Thresholds:")
print("Low threshold    :", q1)
print("Medium threshold :", q2)


def assign_production_category(value):

    if value <= q1:
        return "Low"

    elif value <= q2:
        return "Medium"

    else:
        return "High"


df["Production_Category"] = df[
    "Total_Vegetable_Production"
].apply(assign_production_category)


print("\nProduction Category Distribution:")
print(df["Production_Category"].value_counts())


# ------------------------------------------------
# 12. SAVE PROCESSED DATASET
# ------------------------------------------------

processed_csv_path = os.path.join(
    BASE_DIR,
    "processed_vegetable_data.csv"
)

df.to_csv(
    processed_csv_path,
    index=False
)

print("\nProcessed dataset saved:")
print(processed_csv_path)


# ------------------------------------------------
# 13. SELECT INPUT FEATURES
# ------------------------------------------------

X = df[crop_columns].copy()

y = df["Production_Category"].copy()

print("\nInput Features:")

for feature in crop_columns:
    print("-", feature)


# ------------------------------------------------
# 14. LABEL ENCODING
# ------------------------------------------------

label_encoder = LabelEncoder()

y_encoded = label_encoder.fit_transform(y)

print("\nEncoded Classes:")

for i, class_name in enumerate(label_encoder.classes_):
    print(i, "->", class_name)


# ------------------------------------------------
# 15. SAVE LABEL ENCODER
# ------------------------------------------------

label_encoder_path = os.path.join(
    BASE_DIR,
    "label_encoder.pkl"
)

with open(label_encoder_path, "wb") as file:

    pickle.dump(
        label_encoder,
        file
    )

print("\nLabel Encoder saved:")
print(label_encoder_path)


# ------------------------------------------------
# 16. TRAIN-TEST SPLIT
# ------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_encoded,
    test_size=0.20,
    random_state=42,
    stratify=y_encoded
)

print("\nTrain-Test Split:")
print("Training samples:", len(X_train))
print("Testing samples :", len(X_test))


# ================================================================
# RANDOM FOREST MODEL
# ================================================================


# ------------------------------------------------
# 17. TRAIN RANDOM FOREST
# ------------------------------------------------

rf_model = RandomForestClassifier(
    n_estimators=300,
    max_depth=None,
    min_samples_split=2,
    min_samples_leaf=1,
    random_state=42,
    class_weight="balanced"
)

print("\nTraining Random Forest...")

rf_model.fit(
    X_train,
    y_train
)

print("Random Forest training completed.")


# ------------------------------------------------
# 18. RANDOM FOREST PREDICTION
# ------------------------------------------------

rf_predictions = rf_model.predict(X_test)

rf_accuracy = accuracy_score(
    y_test,
    rf_predictions
)

print("\nRandom Forest Accuracy:")
print(rf_accuracy)

print("\nRandom Forest Classification Report:")

print(
    classification_report(
        y_test,
        rf_predictions,
        target_names=label_encoder.classes_,
        zero_division=0
    )
)


# ------------------------------------------------
# 19. RANDOM FOREST CONFUSION MATRIX
# ------------------------------------------------

rf_cm = confusion_matrix(
    y_test,
    rf_predictions
)

print("\nRandom Forest Confusion Matrix:")
print(rf_cm)


# ------------------------------------------------
# 20. SAVE RANDOM FOREST MODEL AS PKL
# ------------------------------------------------

rf_model_path = os.path.join(
    BASE_DIR,
    "vegetable_rf_model.pkl"
)

with open(rf_model_path, "wb") as file:

    pickle.dump(
        rf_model,
        file
    )

print("\nRandom Forest model saved:")
print(rf_model_path)


# ================================================================
# STANDARD SCALING FOR NEURAL NETWORK
# ================================================================


# ------------------------------------------------
# 21. SCALE FEATURES
# ------------------------------------------------

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)


# ------------------------------------------------
# 22. SAVE SCALER
# ------------------------------------------------

scaler_path = os.path.join(
    BASE_DIR,
    "scaler.pkl"
)

with open(scaler_path, "wb") as file:

    pickle.dump(
        scaler,
        file
    )

print("\nScaler saved:")
print(scaler_path)


# ================================================================
# NEURAL NETWORK
# ================================================================


# ------------------------------------------------
# 23. ONE-HOT ENCODE TARGET
# ------------------------------------------------

number_of_classes = len(
    label_encoder.classes_
)

y_train_nn = to_categorical(
    y_train,
    num_classes=number_of_classes
)

y_test_nn = to_categorical(
    y_test,
    num_classes=number_of_classes
)


# ------------------------------------------------
# 24. CREATE NEURAL NETWORK
# ------------------------------------------------

nn_model = Sequential([
    
    Input(
        shape=(X_train_scaled.shape[1],)
    ),

    Dense(
        64,
        activation="relu"
    ),

    Dropout(0.20),

    Dense(
        32,
        activation="relu"
    ),

    Dropout(0.10),

    Dense(
        16,
        activation="relu"
    ),

    Dense(
        number_of_classes,
        activation="softmax"
    )
])


# ------------------------------------------------
# 25. COMPILE MODEL
# ------------------------------------------------

nn_model.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

print("\nNeural Network Architecture:")

nn_model.summary()


# ------------------------------------------------
# 26. EARLY STOPPING
# ------------------------------------------------

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=15,
    restore_best_weights=True
)


# ------------------------------------------------
# 27. TRAIN NEURAL NETWORK
# ------------------------------------------------

print("\nTraining Neural Network...")

history = nn_model.fit(
    X_train_scaled,
    y_train_nn,

    validation_split=0.20,

    epochs=150,

    batch_size=16,

    callbacks=[
        early_stopping
    ],

    verbose=1
)

print("\nNeural Network training completed.")


# ------------------------------------------------
# 28. NEURAL NETWORK EVALUATION
# ------------------------------------------------

nn_loss, nn_accuracy = nn_model.evaluate(
    X_test_scaled,
    y_test_nn,
    verbose=0
)

print("\nNeural Network Test Loss:")
print(nn_loss)

print("\nNeural Network Test Accuracy:")
print(nn_accuracy)


# ------------------------------------------------
# 29. NEURAL NETWORK PREDICTIONS
# ------------------------------------------------

nn_probabilities = nn_model.predict(
    X_test_scaled,
    verbose=0
)

nn_predictions = np.argmax(
    nn_probabilities,
    axis=1
)

print("\nNeural Network Classification Report:")

print(
    classification_report(
        y_test,
        nn_predictions,
        target_names=label_encoder.classes_,
        zero_division=0
    )
)


# ------------------------------------------------
# 30. SAVE NEURAL NETWORK AS H5
# ------------------------------------------------

h5_model_path = os.path.join(
    BASE_DIR,
    "vegetable_nn_model.h5"
)

nn_model.save(
    h5_model_path
)

print("\nNeural Network H5 model saved:")
print(h5_model_path)


# ================================================================
# FEATURE IMPORTANCE
# ================================================================


# ------------------------------------------------
# 31. RANDOM FOREST FEATURE IMPORTANCE
# ------------------------------------------------

feature_importance = pd.DataFrame({

    "Feature": crop_columns,

    "Importance": rf_model.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

print("\nFeature Importance:")

print(feature_importance)


# ------------------------------------------------
# 32. SAVE FEATURE IMPORTANCE CSV
# ------------------------------------------------

feature_importance_path = os.path.join(
    BASE_DIR,
    "feature_importance.csv"
)

feature_importance.to_csv(
    feature_importance_path,
    index=False
)


# ================================================================
# YAML CONFIGURATION FILE
# ================================================================


# ------------------------------------------------
# 33. CREATE YAML CONFIG
# ------------------------------------------------

yaml_config = {

    "project": {
        "name": "AI-Based Vegetable Production Pattern Analysis",

        "dataset":
            "Production_of_Important_Vegetable_crops_In_Tonnes_2019_20.csv",

        "dataset_rows":
            int(df.shape[0]),

        "dataset_columns":
            int(df.shape[1])
    },

    "features": crop_columns,

    "target": {
        "name":
            "Production_Category",

        "classes":
            label_encoder.classes_.tolist(),

        "low_threshold":
            float(q1),

        "medium_threshold":
            float(q2)
    },

    "random_forest": {
        "algorithm":
            "RandomForestClassifier",

        "number_of_trees":
            300,

        "random_state":
            42,

        "test_accuracy":
            float(rf_accuracy)
    },

    "neural_network": {
        "input_features":
            len(crop_columns),

        "hidden_layers": [
            64,
            32,
            16
        ],

        "output_classes":
            number_of_classes,

        "optimizer":
            "adam",

        "loss":
            "categorical_crossentropy",

        "test_accuracy":
            float(nn_accuracy)
    }
}


# ------------------------------------------------
# 34. SAVE YAML FILE
# ------------------------------------------------

yaml_path = os.path.join(
    BASE_DIR,
    "vegetable_project_config.yaml"
)

with open(
    yaml_path,
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        yaml_config,
        file,
        sort_keys=False,
        allow_unicode=True
    )

print("\nYAML configuration saved:")
print(yaml_path)


# ================================================================
# JSON METADATA
# ================================================================


# ------------------------------------------------
# 35. CREATE JSON METADATA
# ------------------------------------------------

metadata = {

    "project_name":
        "AI-Based Vegetable Production Pattern Analysis",

    "dataset": {
        "file":
            CSV_PATH,

        "rows":
            int(df.shape[0]),

        "columns":
            int(df.shape[1])
    },

    "taluk_column":
        TALUK_COLUMN,

    "input_features":
        crop_columns,

    "number_of_features":
        len(crop_columns),

    "target":
        "Production_Category",

    "production_classes":
        label_encoder.classes_.tolist(),

    "production_thresholds": {
        "low_maximum":
            float(q1),

        "medium_maximum":
            float(q2),

        "high_minimum":
            float(q2)
    },

    "training": {

        "training_samples":
            int(len(X_train)),

        "testing_samples":
            int(len(X_test)),

        "test_size":
            0.20,

        "random_state":
            42
    },

    "model_performance": {

        "random_forest_accuracy":
            float(rf_accuracy),

        "neural_network_accuracy":
            float(nn_accuracy),

        "neural_network_loss":
            float(nn_loss)
    },

    "models": {

        "random_forest":
            "vegetable_rf_model.pkl",

        "neural_network":
            "vegetable_nn_model.h5",

        "scaler":
            "scaler.pkl",

        "label_encoder":
            "label_encoder.pkl"
    },

    "feature_importance":
        feature_importance.to_dict(
            orient="records"
        )
}


# ------------------------------------------------
# 36. SAVE JSON FILE
# ------------------------------------------------

json_path = os.path.join(
    BASE_DIR,
    "vegetable_model_metadata.json"
)

with open(
    json_path,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        metadata,
        file,
        indent=4
    )

print("\nJSON metadata saved:")
print(json_path)


# ================================================================
# SAVE TRAINING HISTORY
# ================================================================


# ------------------------------------------------
# 37. TRAINING HISTORY CSV
# ------------------------------------------------

history_df = pd.DataFrame(
    history.history
)

history_path = os.path.join(
    BASE_DIR,
    "neural_network_training_history.csv"
)

history_df.to_csv(
    history_path,
    index=False
)


# ================================================================
# FINAL SUMMARY
# ================================================================

print("\n")
print("=" * 70)
print("PROJECT COMPLETED SUCCESSFULLY")
print("=" * 70)

print("\nRandom Forest Accuracy:")
print(
    f"{rf_accuracy * 100:.2f}%"
)

print("\nNeural Network Accuracy:")
print(
    f"{nn_accuracy * 100:.2f}%"
)

print("\nFiles generated:")

generated_files = [

    processed_csv_path,

    rf_model_path,

    h5_model_path,

    scaler_path,

    label_encoder_path,

    yaml_path,

    json_path,

    feature_importance_path,

    history_path
]

for file in generated_files:

    print("✓", file)

print("\nAll files saved inside:")

print(BASE_DIR)

print("\nDone!")
```
